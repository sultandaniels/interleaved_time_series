# Restart Memorization

Load the ortho_haar_big model and train data for irrelevant tokens experiments.

```python
import os
import pickle
import torch
from core.config import Config
from data_train import set_config_params
from models import GPT2
from linalg_helpers import print_matrix
```

```python
# Load ortho_haar_big model
config = Config()
output_dir, ckpt_dir, experiment_name = set_config_params(config, "ortho_haar_big")
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

ckpt_step = 135000  # Adjust as needed - use a checkpoint that exists in your run
ckpt_path = os.path.join(ckpt_dir, "checkpoints", f"step={ckpt_step}.ckpt")
config.override("ckpt_path", ckpt_path)

model = GPT2.load_from_checkpoint(
    config.ckpt_path,
    n_dims_in=config.n_dims_in, n_positions=config.n_positions,
    n_dims_out=config.n_dims_out, n_embd=config.n_embd,
    n_layer=config.n_layer, n_head=config.n_head, use_pos_emb=config.use_pos_emb,
    map_location=device, strict=True
).eval().to(device)

print(f"Loaded ortho_haar_big model from {ckpt_path}")
```

```python
# Load train data: new hay insert and identical haystack
new_hay_insert_path = "/data/shared/ICL_Kalman_Experiments/train_and_test_data/ortho_haar/train_irrelevant_tokens_new_hay_insert_interleaved_traces_ortho_haar_ident_C_haystack_len_1_state_dim_5.pkl"
repeat_haystack_path = "/data/shared/ICL_Kalman_Experiments/train_and_test_data/ortho_haar/train_irrelevant_tokens_new_hay_insert_repeat_haystack_interleaved_traces_ortho_haar_ident_C_haystack_len_1_state_dim_5.pkl"

with open(new_hay_insert_path, "rb") as f:
    train_new_hay_insert = pickle.load(f)

with open(repeat_haystack_path, "rb") as f:
    train_repeat_haystack = pickle.load(f)

print(f"train_new_hay_insert keys: {train_new_hay_insert.keys() if isinstance(train_new_hay_insert, dict) else type(train_new_hay_insert)}")
print(f"train_repeat_haystack keys: {train_repeat_haystack.keys() if isinstance(train_repeat_haystack, dict) else type(train_repeat_haystack)}")

#print shapes of train_new_hay_insert and train_identical_haystack multi_sys_ys
print(f"train_new_hay_insert multi_sys_ys shape: {train_new_hay_insert['multi_sys_ys'].shape}")
print(f"train_repeat_haystack multi_sys_ys shape: {train_repeat_haystack['multi_sys_ys'].shape}")
```

## Look at outputs of model for a single trace

```python
#get a single trace from train_repeat_haystack
single_trace_repeat_haystack = train_repeat_haystack['multi_sys_ys'][39000,0,0,:]
#print shape of single trace
print(f"single trace shape: {single_trace_repeat_haystack.shape}")
#print the single trace using print_matrix
print_matrix(single_trace_repeat_haystack, "repeat haystack trace")


```

```python
#get a single trace from train_identical_haystack
single_trace_new_hay_insert = train_new_hay_insert['multi_sys_ys'][0,0,0,:]
#print shape of single trace
print(f"single trace shape: {single_trace_new_hay_insert.shape}")
#print the single trace using print_matrix
print_matrix(single_trace_new_hay_insert, "new hay insert trace")
```

```python
#run inference on single_trace_repeat_haystack
# Model expects input_dict with "current" key; "current" = input without last timestep (segments[:-1])
current = single_trace_repeat_haystack[:-1, :]  # shape (24, 57)
current_batch = torch.from_numpy(current).float().unsqueeze(0).to(device)  # shape (1, 24, 57)

with torch.no_grad():
    _, intermediate_dict = model.predict_step({"current": current_batch})
    single_trace_repeat_haystack_pred = intermediate_dict["preds"].squeeze(0).cpu().numpy()  # shape (24, 5)

#print the prediction using print_matrix
print_matrix(single_trace_repeat_haystack_pred, "repeat haystack trace prediction")


```
